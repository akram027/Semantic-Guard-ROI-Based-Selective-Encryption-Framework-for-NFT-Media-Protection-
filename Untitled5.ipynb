```python
# ============================================================
# CELL 7: MASK R-CNN ROI DETECTION
# ============================================================

# Transform input
# Convert img_rgb (numpy array) to PIL Image before transforming
input_tensor = mask_transform(Image.fromarray(img_rgb))

input_tensor = input_tensor.to(DEVICE)

# Run inference
with torch.inference_mode():

    start_time = time.perf_counter()

    prediction = mask_model(
        [input_tensor]
    )[0]

    maskrcnn_time = (
        time.perf_counter() - start_time
    )

# Confidence threshold from methodology
CONF_THRESHOLD = 0.70

keep = (
    prediction["scores"]
    >= CONF_THRESHOLD
)

# Extract results
boxes = (
    prediction["boxes"][keep]
    .detach()
    .cpu()
    .numpy()
)

labels = (
    prediction["labels"][keep]
    .detach()
    .cpu()
    .numpy()
)

scores = (
    prediction["scores"][keep]
    .detach()
    .cpu()
    .numpy()
)

masks = (
    prediction["masks"][keep, 0]
    .detach()
    .cpu()
    .numpy()
)

# Convert masks to binary
maskrcnn_binary = (
    np.any(
        masks >= 0.50,
        axis=0
    )
    .astype(np.uint8)
)

# Resize to original image size
maskrcnn_binary = cv2.resize(
    maskrcnn_binary,
    (W, H),
    interpolation=cv2.INTER_NEAREST
)

print(
    "Mask R-CNN inference time:",
    round(maskrcnn_time, 4),
    "seconds"
)

print(
    "Detected objects:",
    len(scores)
)

for i, (label, score) in enumerate(
    zip(labels, scores)
):

    print(
        i,
        COCO_CLASSES[int(label)],
        "confidence =",
        round(float(score), 4)
    )
```

```python
# ============================================================
# CELL 9: LOAD VISION TRANSFORMER
# ============================================================

VIT_NAME = "google/vit-base-patch16-224"

processor = (
    ViTImageProcessor
    .from_pretrained(VIT_NAME)
)

vit_model = (
    ViTModel
    .from_pretrained(
        VIT_NAME,
        output_attentions=True
    )
)

vit_model = vit_model.to(DEVICE)

vit_model.eval()

print("Vision Transformer loaded.")
print("Model:", VIT_NAME)
print("Device:", DEVICE)
```

```python
# ============================================================
# CELL 10: GENERATE VIT ATTENTION MAP
# ============================================================

vit_inputs = processor(
    images=Image.fromarray(img_rgb),
    return_tensors="pt"
)

vit_inputs = {
    key: value.to(DEVICE)
    for key, value in vit_inputs.items()
}

with torch.inference_mode():

    start_time = time.perf_counter()

    vit_output = vit_model(
        **vit_inputs,
        output_attentions=True
    )

    vit_time = (
        time.perf_counter()
        - start_time
    )

# Final transformer layer
attention = (
    vit_output
    .attentions[-1]
)

# Average attention heads
cls_attention = (
    attention[:, :, 0, 1:]
    .mean(dim=1)[0]
)

num_patches = (
    cls_attention.numel()
)

grid_size = int(
    round(
        math.sqrt(num_patches)
    )
)

if grid_size * grid_size != num_patches:
    raise ValueError(
        "Unexpected ViT patch structure."
    )

# Convert patch attention into 2D map
vit_small = (
    cls_attention
    .reshape(
        grid_size,
        grid_size
    )
    .detach()
    .cpu()
    .numpy()
)

# Resize to original image dimensions
vit_map = cv2.resize(
    vit_small,
    (W, H),
    interpolation=cv2.INTER_CUBIC
)

# Normalize 0-1
vit_map = (
    vit_map - vit_map.min()
) / (
    vit_map.max()
    - vit_map.min()
    + 1e-8
)

print(
    "ViT inference time:",
    round(vit_time, 4),
    "seconds"
)

print(
    "Attention map shape:",
    vit_map.shape
)
```

```python
# ============================================================
# CELL 12: MASK R-CNN + VIT FUSION
# ============================================================

FUSION_THRESHOLD = 0.50

# Fusion operation
fused_score = (
    maskrcnn_binary.astype(
        np.float32
    )
    * vit_map
)

# Normalize
fused_score = (
    fused_score
    / (
        fused_score.max()
        + 1e-8
    )
)

# Convert to binary
fused_binary = (
    fused_score
    >= FUSION_THRESHOLD
).astype(np.uint8)

# Remove very small components
num_labels, labels_cc, stats, centroids = (
    cv2.connectedComponentsWithStats(
        fused_binary,
        8
    )
)

MIN_COMPONENT_AREA = max(
    16,
    int(0.00005 * H * W)
)

clean_mask = np.zeros_like(
    fused_binary
)

for i in range(
    1,
    num_labels
):

    area = stats[
        i,
        cv2.CC_STAT_AREA
    ]

    if area >= MIN_COMPONENT_AREA:

        clean_mask[
            labels_cc == i
        ] = 1

fused_binary = clean_mask

# Calculate ROI percentage
roi_ratio = float(
    fused_binary.mean()
)

print(
    f"ROI ratio = {roi_ratio:.4f}"
)

print(
    f"ROI percentage = {roi_ratio*100:.2f}%"
)
```

```python
# ============================================================
# CELL 15: SEGMENTATION METRICS
# ============================================================

def load_binary_mask(
    path,
    target_height,
    target_width
):

    mask = cv2.imread(
        path,
        cv2.IMREAD_GRAYSCALE
    )

    if mask is None:
        raise ValueError(
            "Ground-truth mask could not be read."
        )

    mask = cv2.resize(
        mask,
        (
            target_width,
            target_height
        ),
        interpolation=cv2.INTER_NEAREST
    )

    return (
        mask >= 128
    ).astype(np.uint8)


def calculate_metrics(
    prediction,
    ground_truth
):

    prediction = (
        prediction.astype(bool)
    )

    ground_truth = (
        ground_truth.astype(bool)
    )

    intersection = np.logical_and(
        prediction,
        ground_truth
    ).sum()

    union = np.logical_or(
        prediction,
        ground_truth
    ).sum()

    pred_area = prediction.sum()
    gt_area = ground_truth.sum()

    iou_fg = (
        intersection
        /
        (union + 1e-12)
    )

    dice = (
        2 * intersection
        /
        (pred_area + gt_area + 1e-12)
    )

    # Background IoU
    background_intersection = (
        np.logical_and(
            ~prediction,
            ~ground_truth
        ).sum()
    )

    background_union = (
        np.logical_or(
            ~prediction,
            ~ground_truth
        ).sum()
    )

    iou_bg = (
        background_intersection
        /
        (background_union + 1e-12)
    )

    miou = (
        iou_fg + iou_bg
    ) / 2

    return {
        "IoU": iou_fg,
        "mIoU": miou,
        "Dice": dice,
        "Predicted_ROI_%":
            prediction.mean() * 100,
        "GroundTruth_ROI_%":
            ground_truth.mean() * 100
    }


if GT_PATH is not None:

    ground_truth = load_binary_mask(
        GT_PATH,
        H,
        W
    )

    maskrcnn_metrics = (
        calculate_metrics(
            maskrcnn_binary,
            ground_truth
        )
    )

    fusion_metrics = (
        calculate_metrics(
            fused_binary,
            ground_truth
        )
    )

    metrics_df = pd.DataFrame([
        {
            "Method": "Mask R-CNN",
            **maskrcnn_metrics
        },
        {
            "Method":
                "Mask R-CNN + ViT Fusion",
            **fusion_metrics
        }
    ])

    display(metrics_df)

else:

    print(
        "Ground-truth mask not supplied."
    )

    print(
        "IoU, mIoU and Dice cannot "
        "be experimentally calculated."
    )
```

```python
# ============================================================
# CELL 16: AES-256 FUNCTIONS
# ============================================================

def aes256_encrypt(
    data,
    key,
    nonce
):

    cipher = Cipher(
        algorithms.AES(key),
        modes.CTR(nonce)
    )

    encryptor = cipher.encryptor()

    return (
        encryptor.update(data)
        +
        encryptor.finalize()
    )


def aes256_decrypt(
    data,
    key,
    nonce
):

    cipher = Cipher(
        algorithms.AES(key),
        modes.CTR(nonce)
    )

    decryptor = cipher.decryptor()

    return (
        decryptor.update(data)
        +
        decryptor.finalize()
    )
```

```python
# ============================================================
# CELL 17: SELECTIVE AES-256 ENCRYPTION
# ============================================================

def selective_encrypt(
    image,
    roi_mask
):

    # 256-bit AES key
    key = secrets.token_bytes(32)

    # AES CTR nonce
    nonce = secrets.token_bytes(16)

    encrypted = image.copy()

    roi_pixels = (
        image[
            roi_mask.astype(bool)
        ]
    )

    roi_bytes = (
        roi_pixels.tobytes()
    )

    encrypted_bytes = (
        aes256_encrypt(
            roi_bytes,
            key,
            nonce
        )
    )

    encrypted_pixels = (
        np.frombuffer(
            encrypted_bytes,
            dtype=np.uint8
        )
        .reshape((-1, 3))
    )

    encrypted[
        roi_mask.astype(bool)
    ] = encrypted_pixels

    return (
        encrypted,
        key,
        nonce
    )


start = time.perf_counter()

encrypted_image, aes_key, aes_nonce = (
    selective_encrypt(
        img_rgb,
        fused_binary
    )
)

selective_encryption_time = (
    time.perf_counter()
    - start
)

print(
    "AES-256 key size:",
    len(aes_key) * 8,
    "bits"
)

print(
    "ROI pixels:",
    int(fused_binary.sum())
)

print(
    "Selective encryption time:",
    selective_encryption_time,
    "seconds"
)

Image.fromarray(
    encrypted_image
).save(
    OUT / "selective_encrypted.png"
)

plt.figure(figsize=(12, 7))

plt.imshow(
    encrypted_image
)

plt.axis("off")

plt.title(
    "AES-256 Selectively Encrypted ROI"
)

plt.show()
```

```python
# ============================================================
# CELL 18: SELECTIVE AES-256 DECRYPTION
# ============================================================

def selective_decrypt(
    encrypted_image,
    roi_mask,
    key,
    nonce
):

    decrypted = (
        encrypted_image.copy()
    )

    encrypted_pixels = (
        encrypted_image[
            roi_mask.astype(bool)
        ]
    )

    encrypted_bytes = (
        encrypted_pixels.tobytes()
    )

    original_bytes = (
        aes256_decrypt(
            encrypted_bytes,
            key,
            nonce
        )
    )

    original_pixels = (
        np.frombuffer(
            original_bytes,
            dtype=np.uint8
        )
        .reshape((-1, 3))
    )

    decrypted[
        roi_mask.astype(bool)
    ] = original_pixels

    return decrypted


start = time.perf_counter()

decrypted_image = (
    selective_decrypt(
        encrypted_image,
        fused_binary,
        aes_key,
        aes_nonce
    )
)

decryption_time = (
    time.perf_counter()
    - start
)

exact_recovery = np.array_equal(
    img_rgb,
    decrypted_image
)

print(
    "Decryption time:",
    decryption_time,
    "seconds"
)

print(
    "Exact original recovery:",
    exact_recovery
)

Image.fromarray(
    decrypted_image
).save(
    OUT / "decrypted.png"
)
```

```python
# ============================================================
# CELL 19: FULL-FRAME AES-256 BASELINE
# ============================================================

def full_frame_encrypt(
    image
):

    key = secrets.token_bytes(32)
    nonce = secrets.token_bytes(16)

    raw_bytes = image.tobytes()

    encrypted_bytes = (
        aes256_encrypt(
            raw_bytes,
            key,
            nonce
        )
    )

    return (
        encrypted_bytes,
        key,
        nonce
    )


# Warm-up
full_frame_encrypt(img_rgb)

# Timing
start = time.perf_counter()

full_ciphertext, full_key, full_nonce = (
    full_frame_encrypt(img_rgb)
)

full_encryption_time = (
    time.perf_counter()
    - start
)

full_bytes = img_rgb.size

roi_bytes = (
    int(fused_binary.sum())
    * 3
)

time_reduction = (
    100
    *
    (
        1 -
        selective_encryption_time
        /
        max(
            full_encryption_time,
            1e-12
        )
    )
)

print(
    "Full-frame bytes:",
    full_bytes
)

print(
    "ROI bytes:",
    roi_bytes
)

print(
    "ROI data percentage:",
    roi_bytes / full_bytes * 100
)

print(
    "Full encryption time:",
    full_encryption_time
)

print(
    "Selective encryption time:",
    selective_encryption_time
)

print(
    "Measured time reduction:",
    time_reduction,
    "%"
)
```

```python
# ============================================================
# CELL 20: GENERATE BINARY WATERMARK
# ============================================================

OWNER_DID = (
    "did:example:semantic-guard-owner"
)

# SHA-256 of owner DID
did_hash = hashlib.sha256(
    OWNER_DID.encode()
).digest()

# Convert hash to binary
watermark_bits = np.unpackbits(
    np.frombuffer(
        did_hash,
        dtype=np.uint8
    )
)

print(
    "Watermark length:",
    len(watermark_bits),
    "bits"
)

print(
    "Owner DID:",
    OWNER_DID
)

# Visualization
watermark_display = (
    watermark_bits.reshape(8, 32)
)

plt.figure(figsize=(12, 3))

plt.imshow(
    watermark_display,
    cmap="gray",
    interpolation="nearest"
)

plt.axis("off")

plt.title(
    "Binary Watermark Derived from Owner DID"
)

plt.show()
```

```python
# ============================================================
# CELL 21: DCT WATERMARK EMBEDDING
# ============================================================

def dct_embed(
    image,
    roi_mask,
    bits,
    strength=8.0
):

    ycrcb = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2YCrCb
    ).astype(np.float32)

    Y = ycrcb[:, :, 0].copy()

    positions = []

    for y in range(
        0,
        H - 7,
        8
    ):

        for x in range(
            0,
            W - 7,
            8
        ):

            block_mask = (
                roi_mask[
                    y:y+8,
                    x:x+8
                ]
            )

            # Require most of block to be ROI
            if block_mask.mean() < 0.75:
                continue

            positions.append(
                (y, x)
            )

            if len(positions) >= len(bits):
                break

        if len(positions) >= len(bits):
            break

    if len(positions) < len(bits):

        raise ValueError(
            "ROI does not contain enough "
            "8x8 blocks for watermark."
        )

    for bit, (y, x) in zip(
        bits,
        positions
    ):

        block = Y[
            y:y+8,
            x:x+8
        ]

        dct_block = cv2.dct(
            block
        )

        # Mid-frequency coefficients
        c1 = (3, 4)
        c2 = (4, 3)

        if bit == 1:

            dct_block[c1] += strength
            dct_block[c2] -= strength

        else:

            dct_block[c1] -= strength
            dct_block[c2] += strength

        Y[
            y:y+8,
            x:x+8
        ] = cv2.idct(
            dct_block
        )

    ycrcb[:, :, 0] = (
        np.clip(
            Y,
            0,
            255
        )
    )

    output = cv2.cvtColor(
        ycrcb.astype(np.uint8),
        cv2.COLOR_YCrCb2RGB
    )

    return output, positions
```

```python
# ============================================================
# CELL 22: APPLY DCT WATERMARK
# ============================================================

try:

    watermarked_image, watermark_positions = (
        dct_embed(
            img_rgb,
            fused_binary,
            watermark_bits,
            strength=8.0
        )
    )

    print(
        "Watermark embedded successfully."
    )

    print(
        "Number of watermark blocks:",
        len(watermark_positions)
    )

except ValueError as error:

    print(
        "Watermark embedding failed:"
    )

    print(error)

    watermarked_image = img_rgb.copy()

    watermark_positions = []


Image.fromarray(
    watermarked_image
).save(
    OUT / "watermarked.png"
)

plt.figure(figsize=(12, 7))

plt.imshow(
    watermarked_image
)

plt.axis("off")

plt.title(
    "DCT Watermarked Image"
)

plt.show()
```

```python
# ============================================================
# CELL 23: WATERMARK EXTRACTION
# ============================================================

def dct_extract(
    image,
    positions,
    number_of_bits
):

    ycrcb = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2YCrCb
    ).astype(np.float32)

    Y = ycrcb[:, :, 0]

    recovered_bits = []

    for y, x in positions[
        :number_of_bits
    ]:

        block = Y[
            y:y+8,
            x:x+8
        ]

        dct_block = cv2.dct(
            block
        )

        if (
            dct_block[3,4]
            >=
            dct_block[4,3]
        ):

            recovered_bits.append(1)

        else:

            recovered_bits.append(0)

    return np.array(
        recovered_bits,
        dtype=np.uint8
    )


if len(watermark_positions) > 0:

    extracted_bits = dct_extract(
        watermarked_image,
        watermark_positions,
        len(watermark_bits)
    )

    watermark_accuracy = np.mean(
        extracted_bits
        ==
        watermark_bits
    )

    print(
        "Watermark extraction accuracy:",
        watermark_accuracy
    )

else:

    watermark_accuracy = np.nan

    print(
        "No watermark positions available."
    )
```

```python
# ============================================================
# CELL 24: PSNR AND SSIM
# ============================================================

def calculate_psnr(
    original,
    processed
):

    original = (
        original.astype(np.float64)
    )

    processed = (
        processed.astype(np.float64)
    )

    mse = np.mean(
        (
            original
            -
            processed
        ) ** 2
    )

    if mse == 0:
        return float("inf")

    return (
        10
        *
        np.log10(
            (255 ** 2)
            /
            mse
        )
    )


def calculate_ssim(
    original,
    processed
):

    return structural_similarity(
        original,
        processed,
        channel_axis=2,
        data_range=255
    )


psnr_value = calculate_psnr(
    img_rgb,
    watermarked_image
)

ssim_value = calculate_ssim(
    img_rgb,
    watermarked_image
)

print(
    "PSNR:",
    psnr_value,
    "dB"
)

print(
    "SSIM:",
    ssim_value
)
```

```python
# ============================================================
# CELL 28: ROI-BASED HYBRID COMPRESSION
# ============================================================

def jpeg_attack(
    image,
    quality
):

    # Encode to JPEG in memory
    _, encoded_image = cv2.imencode(
        ".jpg",
        image,
        [cv2.IMWRITE_JPEG_QUALITY, quality]
    )

    # Decode from JPEG in memory
    decoded_image = cv2.imdecode(
        encoded_image,
        cv2.IMREAD_UNCHANGED
    )

    return decoded_image

BACKGROUND_JPEG_QUALITY = 25

compressed_background = jpeg_attack(
    img_rgb,
    BACKGROUND_JPEG_QUALITY
)

# Preserve ROI from original image
# Compress background only
hybrid_image = np.where(
    fused_binary[..., None].astype(bool),
    img_rgb,
    compressed_background
)

# Save original and hybrid images
original_file = (
    OUT / "original.png"
)

hybrid_file = (
    OUT / "hybrid_compressed.jpg"
)

Image.fromarray(
    img_rgb
).save(
    original_file
)

Image.fromarray(
    hybrid_image
).save(
    hybrid_file,
    quality=BACKGROUND_JPEG_QUALITY
)

original_size = (
    os.path.getsize(
        original_file
    )
)

hybrid_size = (
    os.path.getsize(
        hybrid_file
    )
)

file_reduction = (
    100
    *
    (
        1
        -
        hybrid_size
        /
        original_size
    )
)

print(
    "Original file size:",
    original_size,
    "bytes"
)

print(
    "Hybrid file size:",
    hybrid_size,
    "bytes"
)

print(
    "File size reduction:",
    file_reduction,
    "%"
)
```
